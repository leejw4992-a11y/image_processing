```python
import cv2

image = cv2.imread("like_lenna.png", cv2.IMREAD_GRAYSCALE)

if image is not None:
    print("read!")
else:
    print("not!")

print(type(image))

#<class 'numpy.ndarray'> -> 텐서화 되는것.
```

```python
from pathlib import Path
import matplotlib.pyplot as plt

plt.imshow(image, cmap="gray")
plt.title("Like Lenna")
plt.axis("off")
plt.show()
```

```python
print(image.shape)
```

```python
image_small = cv2.resize(image, (100,100))
plt.imshow(image_small, cmap="gray")
plt.title("SMALL Like Lenna")
plt.axis("off")
plt.show()
```

# 상하반전

```python
image_flip = cv2.flip(image,0) #상하반전
plt.imshow(image_flip, cmap="gray")
plt.title("FLIP Like Lenna")
plt.axis("off")
plt.show()
```

# 좌우반전

```python
image_flip = cv2.flip(image,1) #좌우반전
plt.imshow(image_flip, cmap="gray")
plt.title("FLIP Like Lenna")
plt.axis("off")
plt.show()
```

```python
height, width = image.shape
matrix = cv2.getRotationMatrix2D((width / 2, height / 2),90,1) #90도 로테이션하기위해 계산만한것.
res = cv2.warpAffine(image, matrix, (width, height)) #실제로 이미지를 로테이션하게하는것.

plt.imshow(res, cmap="gray")
plt.title("Rotation Like Lenna")
plt.axis("off")
plt.show()
```

```python
# height, width = image.shape
matrix = cv2.getRotationMatrix2D((width / 2, height / 2),30,1) 
res = cv2.warpAffine(image, matrix, (width, height), borderValue = 200) 

plt.imshow(res, cmap="gray")
plt.title("30 Rotation Like Lenna")
plt.axis("off")
plt.show()
```
